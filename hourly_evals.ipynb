# Tübingen

```python
import pandas as pd 

# get hourly weather data for Tübingen (2024-11-01 to 2025-10-31)
tue_weather_hourly = pd.read_csv('weather_data/hourly/tuebingen_weather_2024-11-01_2025-10-31.csv')

# get hourly bike count data for Tübingen (2024-11-01 to 2025-10-31)
tue_bike_hourly = pd.read_csv('eco-counter/all_cities/2024-11-01_2025-10-31_hourly/stadt_tübingen_hourly_2024-11-01_2025-10-31.csv')
tue_bike_hourly = tue_bike_hourly[tue_bike_hourly['counter_site'] == 'Fuß- & Radtunnel Südportal - Derendinger Allee']

# reset index so there is no old index messing things up
tue_weather_hourly = tue_weather_hourly.reset_index(drop=True)
tue_bike_hourly = tue_bike_hourly.reset_index(drop=True)

# now set hour index cleanly from 0 to 8760
tue_weather_hourly['hour'] = range(len(tue_weather_hourly))
tue_bike_hourly['hour'] = range(len(tue_bike_hourly))

tue_weather_hourly = tue_weather_hourly.set_index('hour')
tue_bike_hourly = tue_bike_hourly.set_index('hour')

# sanity check amount of data
print(f'Hourly weather data points: {len(tue_weather_hourly)}')
print(f'Hourly bike data points for Tübingen: {len(tue_bike_hourly)}')
print()


import matplotlib.pyplot as plt
# get relevant columns
tue_rainfall_hourly = tue_weather_hourly['prcp']
tue_temperature_hourly = tue_weather_hourly['temp']
tue_bike_counts_hourly = tue_bike_hourly['channels_all']

# interpret all values 'na' as missing values
tue_rainfall_hourly = pd.to_numeric(tue_rainfall_hourly, errors='coerce')
tue_temperature_hourly = pd.to_numeric(tue_temperature_hourly, errors='coerce')
tue_bike_counts_hourly = pd.to_numeric(tue_bike_counts_hourly, errors='coerce')

# combine into single DataFrame (dropping missing values)
df = pd.concat([
    tue_bike_counts_hourly.rename("bike"),
    tue_rainfall_hourly.rename("rain"),
    tue_temperature_hourly.rename("temp")
], axis=1)

# Align only on shared timestamps
df = df.dropna()
print('Dropping missing values, remaining data points:', len(df))
print(df.corr())

# fill missing values 
tue_rainfall_hourly = tue_rainfall_hourly.ffill()
tue_temperature_hourly = tue_temperature_hourly.ffill()
tue_bike_counts_hourly = tue_bike_counts_hourly.ffill()

df = pd.concat([
    tue_bike_counts_hourly.rename("bike"),
    tue_rainfall_hourly.rename("rain"),
    tue_temperature_hourly.rename("temp")
], axis=1)

print('\nAfter filling missing values, data points:', len(df))
print(df.corr())

```

```python
def calc_hourly_correlation(city_name, bike_file, weather_file, counter_site):
    import pandas as pd 
    print('----------------------------------------------------')
    print(f'City: {city_name}, Counter Site: {counter_site}')
    
    # get hourly weather data
    weather_hourly = pd.read_csv(weather_file)
    # get hourly bike count data
    bike_hourly = pd.read_csv(bike_file)
    bike_hourly = bike_hourly[bike_hourly['counter_site'] == counter_site]
    
    # reset index so there is no old index messing things up
    weather_hourly = weather_hourly.reset_index(drop=True)
    bike_hourly = bike_hourly.reset_index(drop=True)
    # now set hour index cleanly (from 0 to 8760)
    weather_hourly['hour'] = range(len(weather_hourly))
    bike_hourly['hour'] = range(len(bike_hourly))
    weather_hourly = weather_hourly.set_index('hour')
    bike_hourly = bike_hourly.set_index('hour')
    
    # get relevant columns
    rainfall_hourly = weather_hourly['prcp']
    temperature_hourly = weather_hourly['temp']
    bike_counts_hourly = bike_hourly['channels_all']
    
    # interpret all values 'na' as missing values
    rainfall_hourly = pd.to_numeric(rainfall_hourly, errors='coerce')
    temperature_hourly = pd.to_numeric(temperature_hourly, errors='coerce')
    bike_counts_hourly = pd.to_numeric(bike_counts_hourly, errors='coerce')
    
    # Apply EMA to weather data (span hours for smoothing)
    rainfall_ema = rainfall_hourly.ewm(span=4, adjust=False).mean()
    temperature_ema = temperature_hourly.ewm(span=4, adjust=False).mean()
    
    # combine into single DataFrame with EMA values
    df = pd.concat([
        bike_counts_hourly.rename("bike"),
        rainfall_ema.rename("rain_ema"),
        temperature_ema.rename("temp_ema")
    ], axis=1)
    df = df.dropna()
    
    print(f'Data points after dropping missing values: {len(df)}')
    print('\nCorrelations between bike counts and weather EMA:')
    print(f"  Bike vs Rain EMA: {df['bike'].corr(df['rain_ema']):.4f}")
    print(f"  Bike vs Temp EMA: {df['bike'].corr(df['temp_ema']):.4f}")
    print('----------------------------------------------------')

calc_hourly_correlation('Tübingen',
                        'eco-counter/all_cities/2024-11-01_2025-10-31_hourly/stadt_tübingen_hourly_2024-11-01_2025-10-31.csv',
                        'weather_data/hourly/tuebingen_weather_2024-11-01_2025-10-31.csv',
                        'Fuß- & Radtunnel Südportal - Derendinger Allee')
calc_hourly_correlation('Tübingen',
                        'eco-counter/all_cities/2024-11-01_2025-10-31_hourly/stadt_tübingen_hourly_2024-11-01_2025-10-31.csv',
                        'weather_data/hourly/tuebingen_weather_2024-11-01_2025-10-31.csv',
                        'Radbrücke Mitte - Wöhrdstraße')
calc_hourly_correlation('Tübingen',
                        'eco-counter/all_cities/2024-11-01_2025-10-31_hourly/stadt_tübingen_hourly_2024-11-01_2025-10-31.csv',
                        'weather_data/hourly/tuebingen_weather_2024-11-01_2025-10-31.csv',
                        'Unterführung Steinlach/Karlstraße Südseite - Steinlachallee')
calc_hourly_correlation('Tübingen',
                        'eco-counter/all_cities/2024-11-01_2025-10-31_hourly/stadt_tübingen_hourly_2024-11-01_2025-10-31.csv',
                        'weather_data/hourly/tuebingen_weather_2024-11-01_2025-10-31.csv',
                        'Neckartalradweg Hirschau - parallel L371')
calc_hourly_correlation('Heidelberg',
                        'eco-counter/all_cities/2024-11-01_2025-10-31_hourly/stadt_heidelberg_hourly_2024-11-01_2025-10-31.csv',
                        'weather_data/hourly/heidelberg_weather_2024-11-01_2025-10-31.csv',
                        'Mannheimer Straße')
calc_hourly_correlation('Mannheim',
                        'eco-counter/all_cities/2024-11-01_2025-10-31_hourly/stadt_mannheim_hourly_2024-11-01_2025-10-31.csv',
                        'weather_data/hourly/mannheim_weather_2024-11-01_2025-10-31.csv',
                        'Konrad-Adenauer-Brücke')
calc_hourly_correlation('Mannheim',
                        'eco-counter/all_cities/2024-11-01_2025-10-31_hourly/stadt_mannheim_hourly_2024-11-01_2025-10-31.csv',
                        'weather_data/hourly/mannheim_weather_2024-11-01_2025-10-31.csv',
                        'Luzenbergstr.')
```
